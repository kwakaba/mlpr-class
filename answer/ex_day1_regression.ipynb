# Exercise 1

```python
import numpy as np
xs = np.array([110, 140, 80, 150, 130, 90])
ys = np.array([26, 27, 23, 28, 26, 24])

def predict(xs, w, b):
    return w * xs + b

def mse_loss(ys, ys_pred):
    return np.mean((ys - ys_pred) ** 2)

# (1) Calculate MSE loss for w = 0.1 and b = 15
w = 0.1
b = 15
ys_pred = predict(xs, w, b)
loss = mse_loss(ys, ys_pred)
print(f"Loss for w = {w}, b = {b}: {loss:.4f}")

# (2) Calculate MSE loss for w = 0.05 and b = 20
w = 0.05
b = 20
ys_pred = predict(xs, w, b)
loss = mse_loss(ys, ys_pred)
print(f"Loss for w = {w}, b = {b}: {loss:.4f}")
```

```python
xs = np.array([0, 1, 2, 3, 4, 5])
ys = np.array([0, 2, 1, 3, 3, 4])

def mse_gradient(xs, ys, ys_pred):
    n = len(xs)
    dL_dw = (-2/n) * np.sum(xs * (ys - ys_pred))
    dL_db = (-2/n) * np.sum(ys - ys_pred)
    return dL_dw, dL_db

# Calculate the gradients for w = 1 and b = 0
w = 1
b = 0
ys_pred = predict(xs, w, b)
dL_dw, dL_db = mse_gradient(xs, ys, ys_pred)
print(f"Gradient for w = {w}, b = {b}: -dL/dw = {-dL_dw:.4f}, -dL/db = {-dL_db:.4f}")
```

# Exercise 4

```python
x1s = np.array([1, 2, 3, 4, 5, 6])
x2s = np.array([4, 6, 2, 4, 5, 3])
ys = np.array([-2, -3, 2, 1, 1, 4])
N = 6

w = [0.0, 0.0]
b = 0.0
lr = 0.01

for epoch in range(10000):
    # (1) Calculate the loss and gradients
    ys_pred = w[0] * x1s + w[1] * x2s + b
    loss = mse_loss(ys, ys_pred)
    dw0 = (-2/N) * np.sum(x1s * (ys - ys_pred))
    dw1 = (-2/N) * np.sum(x2s * (ys - ys_pred))
    db = (-2/N) * np.sum(ys - ys_pred)
    # (2) Update the parameters
    w[0] -= lr * dw0
    w[1] -= lr * dw1
    b -= lr * db
    # (3) Print the loss and w, b for each epoch
    print(f"epoch: {epoch}, loss: {loss:.4f}, w1: {w[0]:.4f}, w2: {w[1]:.4f}, b: {b:.4f}")
    
print(f"final w1: {w[0]:.4f}, w2: {w[1]:.4f}, b: {b:.4f}")
```
