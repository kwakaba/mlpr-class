# Exercise 1

```python
import numpy as np
xs = np.array([1.2, 2.2, 2.7, 2.8, 3.1, 3.5])
ys = np.array([0, 0, 1, 0, 1, 1])

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict(xs, w, b):
    return sigmoid(w * xs + b)

def bce_loss(ys, ys_pred, verbose=False):
    losses = -(ys * np.log(ys_pred) + (1 - ys) * np.log(1 - ys_pred))
    if verbose:
        for i in range(len(losses)):
            print(f"Loss for sample {i}: {losses[i]:.4f}")
    return losses.mean()

# (1) Calculate BCE loss for w = 3 and b = -6
w = 3
b = -6
ys_pred = predict(xs, w, b)
loss = bce_loss(ys, ys_pred, verbose=True)
print(f"Loss for w = {w}, b = {b}: {loss:.4f}")

# (2) Calculate BCE loss for w = 4 and b = -10
w = 4
b = -10
ys_pred = predict(xs, w, b)
loss = bce_loss(ys, ys_pred, verbose=True)
print(f"Loss for w = {w}, b = {b}: {loss:.4f}")
```

# Exercise 2

```python
def bce_gradient(xs, ys, ys_pred, verbose=False):
    dL_dz = ys_pred - ys
    dL_dw = (dL_dz * xs)
    dL_db = dL_dz
    if verbose:
        for i in range(len(xs)):
            print(f"Gradient for sample {i}: dL/dw = {dL_dw[i]:.4f}, dL/db = {dL_db[i]:.4f}")
    return dL_dw.mean(), dL_db.mean()

# (1) Calculate the gradient for w = 3 and b = -6
w = 3
b = -6
ys_pred = predict(xs, w, b)
dL_dw, dL_db = bce_gradient(xs, ys, ys_pred, verbose=True)
print(f"Gradient for w = {w}, b = {b}: -dL/dw = {-dL_dw:.4f}, -dL/db = {-dL_db:.4f}")

# (2) Calculate the gradient for w = 4 and b = -10
w = 4
b = -10
ys_pred = predict(xs, w, b)
dL_dw, dL_db = bce_gradient(xs, ys, ys_pred, verbose=True)
print(f"Gradient for w = {w}, b = {b}: -dL/dw = {-dL_dw:.4f}, -dL/db = {-dL_db:.4f}")
```

# Exercise 3

```python
x1s = np.array([1, 2, 3, 4, 4, 6])
x2s = np.array([4, 6, 2, 4, 6, 3])
ys = np.array([0, 0, 1, 1, 0, 1])
N = 6

w = [0.0, 0.0]
b = 0.0
lr = 0.1

for epoch in range(10000):
    # (1) Calculate the loss and gradients
    ys_pred = sigmoid(w[0] * x1s + w[1] * x2s + b)
    loss = bce_loss(ys, ys_pred)
    dw0 = (1/N) * np.sum(x1s * (ys_pred - ys))
    dw1 = (1/N) * np.sum(x2s * (ys_pred - ys))
    db = (1/N) * np.sum(ys_pred - ys)
    # (2) Update the parameters
    w[0] -= lr * dw0
    w[1] -= lr * dw1
    b -= lr * db
    # (3) Print the loss and w, b for each epoch
    print(f"epoch: {epoch}, loss: {loss:.4f}, w1: {w[0]:.4f}, w2: {w[1]:.4f}, b: {b:.4f}")
    
print(f"final loss: {loss:.4f}, w1: {w[0]:.4f}, w2: {w[1]:.4f}, b: {b:.4f}")
```
